```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.edge.options import Options
import time
import pandas as pd
 
driver = webdriver.Edge()
 
Searches = ["Data Engineer","Data Science","AI Developer","Machine Learning Engineer"]
list_of_jobs = []
for search in Searches:
    driver.get('https://wuzzuf.net/')
 
    driver.find_element(By.XPATH, "//input[@placeholder='Search Jobs (e.g. Sales in Maadi)']").send_keys(search)

    driver.find_element(By.XPATH,"//button[@type='submit']").click()
    time.sleep(2)
    driver.refresh()
    time.sleep(10)

    for i in range (2):
       time.sleep(4)
       jobs = driver.find_elements(By.CLASS_NAME, 'css-ghe2tq')
       for job in jobs:
        job_title = job.find_element(By.CLASS_NAME, 'css-o171kl').text
        job_company = job.find_element(By.CLASS_NAME, 'css-ipsyv7').text
        job_location = job.find_element(By.CLASS_NAME, 'css-16x61xq').text
        job_tags = job.find_element(By.CLASS_NAME, 'css-5jhz9n').text.replace('\n', ' / ')
        job_experience  = job.find_element(By.CSS_SELECTOR, "div > div.css-1rhj4yg > div:nth-child(2) > span").text
        job_url = job.find_element(By.CLASS_NAME, 'css-o171kl').get_attribute('href')
 
        list_of_jobs.append({
            'job_title': job_title, 
            'job_company': job_company,
            'job_location': job_location,
            'job_tags': job_tags,
            'job_experience': job_experience,
            'job_url': job_url
    })
       
       time.sleep(4)
       driver.find_element( By.CSS_SELECTOR,"button[class='css-kefogq ezfki8j0']").click()
       time.sleep(4)  

df = pd.DataFrame(list_of_jobs)
df.drop_duplicates(subset=['job_url'], inplace=True)
df.to_csv('Wuzzuf.csv', index=False)

```
